## Introduction

The UniPool project aims to develop a ride-pooling web/mobile application for Strathmore University students. The application will require a machine learning component that can make use of transportation-related data.

For this deliverable, a ride-sharing dataset was selected and explored to determine whether it contains relevant information that could be useful for developing the machine learning engine of the UniPool application.

The dataset contains information such as trip locations, distance, fare amount, trip status, payment method, and trip times. These variables are relevant to a ride-pooling application because they describe different characteristics of individual trips.


## Dataset Discovery and Selection

Several open-source datasets related to ride-sharing and transportation were considered for the UniPool project. The datasets were evaluated based on their relevance to ride-pooling, the availability of useful variables, the presence of a potential target/output variable, and their suitability for machine learning.

The selected dataset is the Uber Trips Dataset because it contains 50,000 trip records and includes variables describing trip locations, distance, fare, trip status, payment method, and trip times.


```python
import pandas as pd
```

```python
df = pd.read_csv("uber_trips_dataset_50k.csv")
```

```python
df.head()
```

## a. How many rows and columns are contained in the dataset?


```python
df.shape
```

### Response

The dataset contains **50,000 rows and 14 columns**. The rows represent individual trip records, while the columns contain the different attributes associated with each trip.


## b. What datatypes are contained in the dataset?


```python
df.dtypes
```

### Response

The dataset contains three main datatypes: `int64`, `float64`, and `object`. The `int64` datatype is used for integer values such as the trip, driver, and rider IDs. The `float64` datatype is used for numerical values that can contain decimal values, such as latitude, longitude, distance, and fare. The `object` datatype is used for text or categorical variables such as city, status, payment method, and the recorded trip times.


## c. Is the dataset complete? i.e. no missing values?


```python
df.isnull().sum()
```

```python
df.isnull().sum().sum()
```

### Response

The dataset contains no missing values. The missing-value check returned 0 for every column, and the total number of missing values in the dataset is 0. Therefore, the dataset is complete with respect to missing values.




## d. Slice out the first 15 rows and last 20 rows and merge them into a new dataframe titled `df_sample`.


```python
first_15 = df.head(15)
last_20 = df.tail(20)

df_sample = pd.concat([first_15, last_20])

df_sample
```

```python
df_sample.shape
```

### Response

The first 15 rows and the last 20 rows of the dataset were extracted and combined using `pd.concat()`. The resulting dataframe, `df_sample`, contains **35 rows and 14 columns** and has been displayed above.
